
 2 Create a calculated column discounted price

   -For Task 2, create a **Calculated Column** called **Discounted Price**.

Since:

* **Total Price = Quantity × UnitPrice**
* **Discount** is stored as a decimal (e.g., 0.04 = 4%)

The discounted price is:

[
\text{Discounted Price} = \text{Total Price} \times (1 - \text{Discount})
]

### DAX Formula

If you already created the **Total Price** column:

```DAX
Discounted Price = SalesData[Total Price] * (1 - SalesData[Discount])
```

Or directly:

```DAX
Discounted Price =
SalesData[Quantity] * SalesData[UnitPrice] * (1 - SalesData[Discount])
```

### Example

For OrderID **10001**:

* Quantity = 9
* UnitPrice = 495.78
* Discount = 0.04 (4%)

Total Price:

```text
9 × 495.78 = 4462.02
```

Discounted Price:

```text
4462.02 × (1 - 0.04)
= 4462.02 × 0.96
= 4283.54
```

### Result

| OrderID | Quantity | UnitPrice | Discount | Total Price | Discounted Price |
| ------- | -------- | --------- | -------- | ----------- | ---------------- |
| 10001   | 9        | 495.78    | 0.04     | 4462.02     | 4283.54          |
| 10002   | 8        | 35.45     | 0.29     | 283.60      | 201.36           |
| 10003   | 6        | 442.04    | 0.01     | 2652.24     | 2625.72          |

### Final DAX

```DAX
Discounted Price =
SalesData[Quantity] * SalesData[UnitPrice] * (1 - SalesData[Discount])
```

This column represents the **net order value after applying the discount**.


3 .Create a measure Average Order Value (AOV)

  -### Task 3: Create a Measure – Average Order Value (AOV)

**Average Order Value (AOV)** measures the average revenue generated per order.

### DAX Measure

```DAX
Average Order Value (AOV) =
DIVIDE(
    SUM(SalesData[SalesAmount]),
    DISTINCTCOUNT(SalesData[OrderID])
)
```

### Explanation

* `SUM(SalesData[SalesAmount])` → Total Sales Revenue
* `DISTINCTCOUNT(SalesData[OrderID])` → Total Number of Orders
* `DIVIDE()` → Safely performs division and handles divide-by-zero errors.

### Alternative (using an existing Total Sales measure)

```DAX
Total Sales =
SUM(SalesData[SalesAmount])

Average Order Value (AOV) =
DIVIDE(
    [Total Sales],
    DISTINCTCOUNT(SalesData[OrderID])
)
```

### Business Meaning

* Higher AOV indicates customers are spending more per order.
* Useful for tracking the effectiveness of:

  * Product bundling
  * Upselling and cross-selling
  * Promotional campaigns
  * Sales channels (Online vs Mobile App)

### KPI Card

Add a **Card Visual** in Power BI and use:

```DAX
Average Order Value (AOV)
```

to display the average revenue per order dynamically based on filters such as:

* Region
* Product Category
* Sales Channel
* Date Range

**Final Measure**

```DAX
Average Order Value (AOV) =
DIVIDE(
    SUM(SalesData[SalesAmount]),
    DISTINCTCOUNT(SalesData[OrderID])
)
```


4 Using DAX, create a measure that calculates cumulative Year-to-Date (YTD) Sales based on the Order
Date

  -### Task 4: Create a Measure for Year-to-Date (YTD) Sales

YTD Sales calculates the cumulative sales from the beginning of the year up to the selected date based on **OrderDate**.

### Step 1: Create a Total Sales Measure

```DAX
Total Sales =
SUM(SalesData[SalesAmount])
```

### Step 2: Create the YTD Sales Measure

```DAX
YTD Sales =
TOTALYTD(
    [Total Sales],
    SalesData[OrderDate]
)
```

### Alternative Using DATESYTD

```DAX
YTD Sales =
CALCULATE(
    [Total Sales],
    DATESYTD(SalesData[OrderDate])
)
```

### Recommended Date Table Approach

For best results in Power BI, create a Date table and mark it as a Date Table:

```DAX
DateTable =
CALENDAR(
    MIN(SalesData[OrderDate]),
    MAX(SalesData[OrderDate])
)
```

Then create a relationship:

```text
DateTable[Date] → SalesData[OrderDate]
```

Use:

```DAX
YTD Sales =
TOTALYTD(
    [Total Sales],
    DateTable[Date]
)
```

### How It Works

Suppose sales are:

| Date        | Sales |
| ----------- | ----: |
| 01-Jan-2022 | 1,000 |
| 02-Jan-2022 | 2,000 |
| 03-Jan-2022 | 1,500 |

YTD Sales becomes:

| Date        | Sales | YTD Sales |
| ----------- | ----: | --------: |
| 01-Jan-2022 | 1,000 |     1,000 |
| 02-Jan-2022 | 2,000 |     3,000 |
| 03-Jan-2022 | 1,500 |     4,500 |

### Final Measure

```DAX
YTD Sales =
TOTALYTD(
    [Total Sales],
    SalesData[OrderDate]
)
```

This measure can be used in line charts, KPI cards, and trend reports to show cumulative sales performance throughout the year.


5. Using DAX, create a measure that calculates cumulative Year-to-Date (YTD) Sales based on the Order
Date

  -Task **5** appears to be the same as Task **4** (calculate cumulative Year-to-Date Sales based on `OrderDate`).

### YTD Sales Measure

First, create a Total Sales measure:

```DAX
Total Sales =
SUM(SalesData[SalesAmount])
```

Then create the YTD Sales measure:

```DAX
YTD Sales =
TOTALYTD(
    [Total Sales],
    SalesData[OrderDate]
)
```

### Alternative Using CALCULATE + DATESYTD

```DAX
YTD Sales =
CALCULATE(
    [Total Sales],
    DATESYTD(SalesData[OrderDate])
)
```

### If You Have a Date Table (Best Practice)

```DAX
YTD Sales =
TOTALYTD(
    [Total Sales],
    'Date'[Date]
)
```

### What It Does

* Starts accumulating sales from **January 1st** of the current year.
* Continues adding sales up to the selected date.
* Resets automatically when a new year begins.

Example:

| Order Date  | Sales | YTD Sales |
| ----------- | ----: | --------: |
| 01-Jan-2022 |  1000 |      1000 |
| 02-Jan-2022 |  2000 |      3000 |
| 03-Jan-2022 |  1500 |      4500 |

**Final Answer:**

```DAX
YTD Sales =
TOTALYTD(
    SUM(SalesData[SalesAmount]),
    SalesData[OrderDate]
)
```

This measure returns the cumulative Year-to-Date Sales based on the `OrderDate` column.


6.  Create a calculated column classifying orders as “High Value” if sales > 2000, else “Low Value”.

 -### Task 6: Create a Calculated Column for Order Classification

Classify each order as:

* **High Value** → if `SalesAmount > 2000`
* **Low Value** → if `SalesAmount <= 2000`

### DAX Calculated Column

```DAX
Order Value Category =
IF(
    SalesData[SalesAmount] > 2000,
    "High Value",
    "Low Value"
)
```

### Explanation

* `IF()` checks the condition.
* If `SalesAmount` is greater than 2000, it returns **"High Value"**.
* Otherwise, it returns **"Low Value"**.

### Example

| OrderID | SalesAmount | Order Value Category |
| ------- | ----------: | -------------------- |
| 10001   |     1064.57 | Low Value            |
| 10003   |     4867.82 | High Value           |
| 10009   |     4940.57 | High Value           |
| 10019   |      454.15 | Low Value            |
| 10022   |     4528.69 | High Value           |

### Alternative Using SWITCH

```DAX
Order Value Category =
SWITCH(
    TRUE(),
    SalesData[SalesAmount] > 2000, "High Value",
    "Low Value"
)
```

### Final Answer

```DAX
Order Value Category =
IF(
    SalesData[SalesAmount] > 2000,
    "High Value",
    "Low Value"
)
```

You can use this column in slicers, charts, and KPIs to compare sales performance between **High Value** and **Low Value** orders.


7.Create a measure to calculate total sales from the Online channel only


  -### Task 7: Create a Measure for Total Sales from the Online Channel Only

This measure sums **SalesAmount** only for orders where **SalesChannel = "Online"**.

### DAX Measure

```DAX
Online Sales =
CALCULATE(
    SUM(SalesData[SalesAmount]),
    SalesData[SalesChannel] = "Online"
)
```

### Explanation

* `SUM(SalesData[SalesAmount])` → Calculates total sales.
* `CALCULATE()` → Modifies the filter context.
* `SalesData[SalesChannel] = "Online"` → Filters the data to include only Online orders.

### Alternative Version (Using Existing Total Sales Measure)

If you already have:

```DAX
Total Sales =
SUM(SalesData[SalesAmount])
```

Then:

```DAX
Online Sales =
CALCULATE(
    [Total Sales],
    SalesData[SalesChannel] = "Online"
)
```

### Business Use

This measure can be used to:

* Compare **Online vs Mobile App** sales.
* Build channel performance dashboards.
* Calculate channel contribution percentages.

### Final Measure

```DAX
Online Sales =
CALCULATE(
    SUM(SalesData[SalesAmount]),
    SalesData[SalesChannel] = "Online"
)
```


8 Create a measure called "Date Difference" that calculates the number of days between the latest  Order
Date and today's date using the DATEDIFF function.

  -### Task 8: Create a Measure – Date Difference

This measure calculates the number of days between the **latest OrderDate** in the dataset and **today's date**.

### DAX Measure

```DAX
Date Difference =
DATEDIFF(
    MAX(SalesData[OrderDate]),
    TODAY(),
    DAY
)
```

### Explanation

* `MAX(SalesData[OrderDate])` → Finds the most recent order date in the current filter context.
* `TODAY()` → Returns the current system date.
* `DATEDIFF(..., DAY)` → Calculates the difference in days.

### Alternative (Ignoring Filters)

If you always want the difference from the latest order date in the entire dataset, regardless of slicers or filters:

```DAX
Date Difference =
DATEDIFF(
    CALCULATE(
        MAX(SalesData[OrderDate]),
        ALL(SalesData)
    ),
    TODAY(),
    DAY
)
```

### Example

Suppose:

* Latest OrderDate = **15-May-2023**
* Today's Date = **11-Jun-2026**

Then:

```text
Date Difference = 1123 days
```

### Final Measure

```DAX
Date Difference =
DATEDIFF(
    MAX(SalesData[OrderDate]),
    TODAY(),
    DAY
)
```

This measure is useful for monitoring data freshness and identifying how many days have passed since the most recent order.
